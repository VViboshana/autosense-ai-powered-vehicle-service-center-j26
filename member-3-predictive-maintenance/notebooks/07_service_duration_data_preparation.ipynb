```python
import pandas as pd

data_path = "/content/drive/MyDrive/AutoSenseAI/data/duration/diagnostic_fixes.csv"

df = pd.read_csv(
    data_path,
    sep="|",
    encoding="utf-8-sig"
)

print("Dataset shape:", df.shape)
df.head()
```

```python
print(df.columns.tolist())
```

```python
print(df.isnull().sum())
```

```python
target = "est_labor_hours"

print(df[target].describe())
```

```python
print(df["labor_status"].value_counts(dropna=False))
```

```python
duration_df = df[
    df["est_labor_hours"].notna()
].copy()

print("Usable rows:", duration_df.shape)
```

```python
print(
    duration_df[
        duration_df["est_labor_hours"] <= 0
    ]
)
```

```python
duration_df["labor_minutes"] = (
    duration_df["est_labor_hours"] * 60
)

duration_df[
    ["fix_title", "est_labor_hours", "labor_minutes"]
].head()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    duration_df["est_labor_hours"],
    bins=20
)

plt.title("Estimated Labor Hour Distribution")
plt.xlabel("Labor hours")
plt.ylabel("Number of repair procedures")

plt.show()
```

```python
difficulty_summary = (
    duration_df
    .groupby("difficulty_level")["est_labor_hours"]
    .agg(["count", "mean", "median", "min", "max"])
)

difficulty_summary
```

```python
title_analysis = (
    duration_df
    .groupby("fix_title")["est_labor_hours"]
    .agg(["count", "nunique", "mean", "min", "max"])
)

print("Unique repair titles:", len(title_analysis))

print(
    "Repeated titles:",
    (title_analysis["count"] > 1).sum()
)

print(
    "Repeated titles with different durations:",
    (
        (title_analysis["count"] > 1) &
        (title_analysis["nunique"] > 1)
    ).sum()
)
```

```python
model_features = [
    "fix_title",
    "difficulty_level",
    "probability_rank",
    "est_parts_cost_min_usd",
    "est_parts_cost_max_usd"
]

model_features
```

```python
duration_df["parts_cost_mid_usd"] = (
    duration_df["est_parts_cost_min_usd"] +
    duration_df["est_parts_cost_max_usd"]
) / 2

duration_df["parts_cost_range_usd"] = (
    duration_df["est_parts_cost_max_usd"] -
    duration_df["est_parts_cost_min_usd"]
)
```

```python
clean_columns = [
    "fix_id",
    "code_id",
    "fix_title",
    "probability_rank",
    "difficulty_level",
    "est_parts_cost_min_usd",
    "est_parts_cost_max_usd",
    "parts_cost_mid_usd",
    "parts_cost_range_usd",
    "est_labor_hours",
    "labor_minutes"
]

clean_df = duration_df[clean_columns].copy()

clean_df.head()
```

```python
print("Rows:", len(clean_df))
print("Missing values:")
print(clean_df.isnull().sum())

print(
    "\nUnique repair titles:",
    clean_df["fix_title"].nunique()
)

print(
    "\nTarget range:",
    clean_df["est_labor_hours"].min(),
    "to",
    clean_df["est_labor_hours"].max(),
    "hours"
)
```

```python
output_dir = "/content/drive/MyDrive/AutoSenseAI/data/duration"

os.makedirs(output_dir, exist_ok=True)

output_path = os.path.join(
    output_dir,
    "service_duration_benchmark.csv"
)

clean_df.to_csv(
    output_path,
    index=False
)

print("Saved:", output_path)
```
